```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../Data/data_transformacion_limpieza_bank.csv')
```

```python
import sys
sys.path.append("../SRC")

import sp_limpieza_bank as sp
```

```python
sp.eda_preliminar(df)
```

```python
job_mapping = {
    'admin.': 'administrative',
    'blue-collar' : 'manual',
    'housemaid' : 'manual',
    'technician' : 'technical',
    'services' : 'technical',
    'management' : 'management',
    'entrepreneur' : 'management',
    'self-employed' : 'management',
    'retired' : 'not_working',
    'unemployed' : 'not_working',
    'student' : 'not_working'
}

df['job_grouped'] = df['job'].map(job_mapping) # agrupamos los diferentes trabajos por categorías porque tiene muchos valores con poca frecuencia. 
```

```python
cols = df.columns.tolist()
cols.insert(cols.index('job') + 1, cols.pop(cols.index('job_grouped'))) # movemos 'job_grouped' justo después de 'job'
df = df[cols]
```

```python
df.drop(columns='job', inplace=True) # eliminamos la columna 'job' ya que hemos creado la nueva que lo agrupa por categorías en 'job_grouped'
```

```python
sp.valores_minus(df)
```

```python
df['age'] = df['age'].astype('Int64')
```

```python
quitar_coma = ['cons_price_idx', 'cons_conf_idx', 'euribor3m', 'nr_employed']

sp.cambiar_coma(df,quitar_coma)
```

```python
gestion_nulos = ['cons_price_idx', 'euribor3m', 'nr_employed']

sp.valor_cero(df, gestion_nulos)

print(df[gestion_nulos].isna().sum())
```

```python
cambiar_type = ['cons_price_idx', 'euribor3m', 'nr_employed']

sp.convertir_float(df, cambiar_type)

print(df[cambiar_type].dtypes)
```

```python
redondear_decimal = ['cons_price_idx', 'euribor3m']

sp.round_decimal(df,redondear_decimal)
```

```python
df['nr_employed'] = df['nr_employed'].round(0).astype(int)
```

```python
df['education'] = df['education'].replace({
    'university.degree': 'university degree',
    'high.school': 'high school',
    'basic.9y': 'basic 9y',
    'professional.course': 'professional course',
    'basic.4y': 'basic 4y',
    'basic.6y': 'basic 6y'
})
```

```python
import locale

locale.setlocale(locale.LC_TIME, 'es_ES.UTF-8')

df['date'] = pd.to_datetime(df['date'], format="%d-%B-%Y")

df['year'] = df['date'].dt.year

df['year'] = df['year'].fillna(0).astype(int)

```

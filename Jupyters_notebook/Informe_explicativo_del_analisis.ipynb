# **Informe de análisis de campaña de marketing directo - Banco Portugués**
## **1. Objetivo del análisis**
     
El objetivo de este análisis es evaluar la efectividad de una campaña de marketing directo basada en llamadas telefónicas y comportamiento online, identificando patrones de clientes, métricas clave de engagement, esfuerzo de contacto y potencial de conversión. Esto permite optimizar recursos y diseñar estrategias de segmentación para futuras campañas. 

## **2. Datos analizados**  

   Se trabajó con la unión de dos datasets:
   - bank-additional.csv: información de la campaña de marketing. (Variables como edad, ocupación, duración de llamadas, número de contactos, resultado de campaña.)
   - customer-details.xlsx: información demográfica y comportamiento online. (Variables como ingresos, número de hijos, visitas mensuales al portal web, antigüedad como cliente.)
    
 Se crearon métricas derivadas para capturar **engagement, intensidad de contacto y actividad web ajustada por antigüedad**, entre otras.

## **3. Métricas clave**

| Métrica                                   | Descripción                                                  | Observaciones                                                                            |
|-------------------------------------------|--------------------------------------------------------------|------------------------------------------------------------------------------------------|
| Customer_tenure_days/Customer_tenure_years| Antigüedad del cliente en el banco                           | La media es ~4,1 años; rango 8 días a 8 años. Permite segmentar campañas por antigüedad. |
| High_web_visit_flag                       | Top 25% de clientes según visitas web                        | Solo 22% de clientes están en el top 25%.                                                |
| Web_engagement_score                      | Normalización de visitas web                                 | Media 0,50; distribución equilibrada.                                                    |
| Web_visit_intensity                       | Visitas mensuales ajustadas por antigüedad                   | Media 5,66; outliers extremos hasta 696 visitas.                                         |
| Tenure_vs_campaign_effort                 | Número de llamadas por año de antigüedad                     | Media 0,83; máximo 91,25 → clientes que requieren mucho esfuerzo.                        |
| Contact_effort_ratio                      | Número de contactos por historial de interacciones           | Media 2,42; máximo 56 → clientes “difíciles”.                                            |
| Avg_duration_per_contact                  | Duración promedio de llamadas                                | Media 165 s; máximo 4918 s → outliers que podrían distorsionar medias.                   |
| Engagement_score_total                    | Combinación de engagement web + duración + contactos previos | Media 0,73; permite identificar clientes con alto engagement.                            |
| Overall_interaction_score                 | Engagement total combinado con esfuerzo de campaña           | Media 1,56; outliers hasta 91 indican clientes con alta interacción.                     |
| Web_activity_per_contact                  | Visitas web por contacto                                     | Media 10,5; permite identificar clientes activos que responden a contacto.               |

## **4. Segmentación de clientes (Clustering)**

Se aplicó **KMeans con 4 clusters** sobre métricas de engagement, esfuerzo de contacto y actividad web.

### **4.1 Centroides de clusters (valores estandarizados)**

| Cluster  | Web_engagement        |  Web_visit_intensity      | Tenure_vs_campaign_effort       | Contact_effort_ratio      | Label sugerido          |
|----------|-----------------------|---------------------------|---------------------------------|---------------------------|-------------------------|
| 0        | Alto                  | Bajo-Medio                | Bajo                            | Bajo                      | Top Engagement - Fácil  |
| 1        | Bajo                  | Medio                     | Alto                            | Alto                      | Difícil / Alto esfuerzo |
| 2        | Medio                 | Medio                     | Medio                           | Medio                     | Promedio                |
| 3        | Alto                  | Alto                      | Alto                            | Alto                      | Premium / Muy activo    |

**Insights:**
- Los clusters permiten identificar segmentos prioritarios para la campaña.
- Clientes con alto engagement y bajo esfuerzo son los que ofrecen mayor retorno de inversión.
- Clientes difíciles consumen recursos elevados con baja conversión; conviene reevaluar contacto con ellos. 

## **5. Outliers y distribución**
- Métricas como Web_visit_intensity, Overall_interaction_score y Avg_duration_per_contact presentan valores extremos.
- Recomendaciones: 
  1. Analizar mediana y percentiles para decisiones estratégicas.
  2. Revisar posibles errores de registro o clientes especiales.
  3. Ajustar métricas derivadas para mitigar efecto de outliers en segmentación.

## **6. Insights clave**
**1. Segmentación efectiva:**
- Antigüedad, actividad web y esfuerzo de contacto muestran alta variabilidad.
- Top 20-25% por engagement -> objetivos prioritarios.


**2. Optimización de recursos:**
- Evitar invertir en clientes con alto esfuerzo y bajo engagement.


**3. Impacto del engagement web:**
- Clientes activos en portal web tienen mayor probabilidad de conversión.


**4. Duración de llamadas:**
- Llamadas muy largas no siempre implican mayor conversión; evaluar eficiencia. 


**5. Clusters para estrategia**
- Permiten asignar campañas personalizadas: fácil, promedio, difícil o premium.


## **7. Recomendaciones estratégicas**
**1. Personalizar campañas según cluster:**
- Top Engagement -> contacto mínimo y promociones dirigidas.
- Difíciles -> reducir contactos o usar otros canales.
- Premium -> atención personalizada y productos de mayor valor.


**2. Monitorear fatiga de clientes:**
- Métricas como Contact_effort_ratio y Avg_duration_per_contact ayudan a identificar riesgo de saturación.


**3. Aprovechar comportamiento web:**
- Integrar engagement online y offline para priorizar clientes "calientes".


**4. Visualización para toma de decisiones:**
- Boxplots y scatterplots muestran distribución, outliers y clusters claramente.
